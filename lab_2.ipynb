# Практическая работа 2
### Исходные данные

```python
sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone', 'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan, 50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb', 'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})
products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})
display(sales)
display(products)
```

### Задание 1. Найти общее количество продаж (сумму Quantity) для каждого продукта

```python
sales.groupby('Product')['Quantity'].sum()
```

### Задание 2. Найти среднюю цену для каждого региона

```python
sales.groupby('Region')['Price'].mean()
```

### Задание 3. Заполнить пропуски в столбце Price средним значением

```python
sales['Price'] = sales['Price'].fillna(sales['Price'].mean())
sales
```

### Задание 4. Отсортировать таблицу sales по столбцу Quantity по убыванию

```python
sales.sort_values('Quantity', ascending=False)
```

### Задание 5. Объединить таблицы sales и products по столбцу Product (inner join)

```python
pd.merge(sales, products, on='Product', how='inner')
```
